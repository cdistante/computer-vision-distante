# Recupero dei risultati · Lezione 20

Questo notebook legge la corsa CPU registrata. Non esegue NeRF, 3DGS o VGGT e non sostituisce i tre TODO del notebook studenti. In Colab avviare R02 e caricare Risultati_riferimento_Lezione_20.zip nello stesso runtime. Localmente tenere lo ZIP accanto al notebook.

Nei materiali del corso si usa il **punto decimale** (0.5) e la **virgola seguita da uno spazio** fra componenti di vettori e argomenti, per esempio (1.2, 1, −0.3). I conteggi interi non usano separatori delle migliaia.

```python
# R02
from pathlib import Path
import zipfile,hashlib,json,importlib.util
archive=Path('Risultati_riferimento_Lezione_20.zip')
if not archive.exists():
    try:
        in_colab=importlib.util.find_spec('google.colab') is not None
    except ModuleNotFoundError:
        in_colab=False
    if not in_colab: raise FileNotFoundError('Mettere lo ZIP di riferimento accanto al notebook')
    from google.colab import files
    uploaded=files.upload()
    if archive.name not in uploaded: raise ValueError('Caricare lo ZIP richiesto')
    archive.write_bytes(uploaded[archive.name])
with zipfile.ZipFile(archive) as z:
    manifest=json.loads(z.read('manifest.json'))
    for name,expected in manifest.items():
        assert hashlib.sha256(z.read(name)).hexdigest()==expected,name
    result=json.loads(z.read('risultati_lezione20.json'))
print('Hash verificati. Dati della corsa CPU di riferimento, non una nuova esecuzione.')
print(json.dumps(result,indent=2))
```
